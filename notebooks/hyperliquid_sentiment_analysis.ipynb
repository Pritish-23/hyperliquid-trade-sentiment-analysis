# Data Cleaning & Type Conversion

Cleaning data and converting type of data to make it analysis ready

```python
# Convert 'date' column from string (object) to datetime
fear_greed['date'] = pd.to_datetime(fear_greed['date'])

# Verify conversion
fear_greed.info()
```

```python
# Convert trade timestamp column into datetime
trades['Timestamp IST'] = pd.to_datetime(trades['Timestamp IST'], format='%d-%m-%Y %H:%M')

# Verify conversion
trades.info()
```

```python
# Extract only the date part from datetime columns so both the datasets can be merged efficiently
fear_greed['date_only'] = fear_greed['date'].dt.date

trades['date_only'] = trades['Timestamp IST'].dt.date
```

```python
# Preview results
fear_greed[['date', 'date_only']].head()
```

```python
trades[['Timestamp IST', 'date_only']].head()
```

```python
# Merge trades dataset with fear & greed dataset using daily date

merged_df = pd.merge(
    trades,
    fear_greed,
    on='date_only',
    how='left'
)

# Check merged dataset
merged_df.head()

# Check shape after merge
merged_df.shape
```

```python
# Check for missing values in the merged dataset

merged_df.isnull().sum()
```

```python
# Remove rows where sentiment data is missing
merged_df = merged_df.dropna(subset=['classification'])

# Verify again
merged_df.isnull().sum()
```

# Feature Engineering

Determining whether the trade was profit or loss

If ClosedPnL > 0 then Profit else Loss

```python
# Create profit/loss indicator column
merged_df['is_profit'] = merged_df['Closed PnL'] > 0

# Preview created column
merged_df[['Closed PnL', 'is_profit']].head()
```

Calculating 'Win rate'

Win rate : How often each trader wins

```python
# Calculating win rate for each trader

win_rate = (
    merged_df.groupby('Account')['is_profit'].mean()* 100
    ).reset_index()

# Rename columns
win_rate.columns = ['Account', 'Win Rate %']

# Preview
win_rate.head()
```

Calculating daily profit/loss for each trader.

Profit/Loss = profit or loss made in all the trades for a day

```python
# Calculating daily PnL for each trader

daily_pnl = (
    merged_df.groupby(['Account', 'date_only'])['Closed PnL'].sum().reset_index()
    )

# Preview
daily_pnl.head()
```

Calculating Avg. Trade size of each trader

Avg. Trade size = mean(Size USD)

```python
# Calculate average trade size for each trader

avg_trade_size = (
    merged_df.groupby('Account')['Size USD'].mean().reset_index()
    )

# Rename columns for better readability
avg_trade_size.columns = ['Account', 'Average Trade Size']

# Preview
avg_trade_size.head()
```

Calculating the amount of trades that took place each day

grouping the data by date column and then capturing its size

```python
# Calculate number of trades per day

trades_per_day = (
    merged_df.groupby('date_only').size().reset_index(name='trade_count')
    )

# Preview
trades_per_day.head()
```

Calculating long/short ratio

```python
# long vs short activity

long_short = merged_df['Side'].value_counts()

print(long_short)
```

```python
long_short_ratio = (
    long_short['BUY'] /
    long_short['SELL']
)

print("Long/Short Ratio:", long_short_ratio)
```

# Exploratory Data Analysis(EDA) & Insights

```python
# Count trades under each sentiment category
sentiment_counts = merged_df['classification'].value_counts()

print(sentiment_counts)
```

```python
# Calculate trade frequency by sentiment

trade_frequency = (
    merged_df.groupby('classification')
    .size()
    .reset_index(name='trade_count')
)

print(trade_frequency)
```

```python
# Visualize trade frequency by sentiment

plt.figure(figsize=(8,5))

sns.barplot(
    x='classification',
    y='trade_count',
    data=trade_frequency
)

plt.title("Trade Frequency by Market Sentiment")
plt.xlabel("Market Sentiment")
plt.ylabel("Number of Trades")

plt.savefig("trade_frequency_by_sentiment.png", dpi=300, bbox_inches='tight')

plt.show()
```

**Insight:**

1. Trading activity was highest during Fear conditions, followed by Greed, while Extreme Fear had the lowest participation.
2. This shows how market emotion strongly influences trader participation levels.

```python
# Analyze average PnL across sentiment categories

sentiment_pnl = (
    merged_df.groupby('classification')['Closed PnL']
    .agg(['mean'])
    .reset_index()
)

print(sentiment_pnl)
```

```python
# Visualize average PnL by sentiment

plt.figure(figsize=(8,5))

sns.barplot(
    x='classification',
    y='mean',
    data=sentiment_pnl
)

plt.title("Average PnL by Market Sentiment")
plt.xlabel("Market Sentiment")
plt.ylabel("Average PnL")

plt.savefig("avg_pnl_by_sentiment.png", dpi=300, bbox_inches='tight')

plt.show()
```

**Insights:**

1. Average trader profitability is highest during Extreme Greed conditions and lowest during Extreme Fear and Neutral conditions.
2. This suggests that trader profitability is strongly influenced by market sentiment and volatility conditions.

```python
# Calculate win rate by sentiment

win_sentiment = (
    merged_df.groupby('classification')['is_profit']
    .mean()
    * 100
).reset_index()

# Rename columns
win_sentiment.columns = ['Market Sentiment', 'Win Rate %']

print(win_sentiment)
```

```python
# Visualize win rate by sentiment

plt.figure(figsize=(8,5))

sns.barplot(
    x='Market Sentiment',
    y='Win Rate %',
    data=win_sentiment
)

plt.title("Win Rate by Market Sentiment")
plt.xlabel("Market Sentiment")
plt.ylabel("Win Rate (%)")

plt.savefig("winrate_by_sentiment.png", dpi=300, bbox_inches='tight')

plt.show()
```

**Insights:**

1. Traders achieved the highest win rates during Extreme Greed conditions, while Extreme Fear, Greed and Neutral showed almost equal win rate.
2. Trader consistency improves in strongly bullish sentiment conditions and weakens during highly fearful markets.

```python
# Analyze average trade size by sentiment

trade_size_sentiment = (
    merged_df.groupby('classification')['Size USD']
    .mean()
    .reset_index()
)

# Rename column
trade_size_sentiment.columns = ['Market Sentiment', 'Average Trade Size']

print(trade_size_sentiment)
```

```python
# Visualize average trade size by sentiment

plt.figure(figsize=(8,5))

sns.barplot(
    x='Market Sentiment',
    y='Average Trade Size',
    data=trade_size_sentiment
)

plt.title("Average Trade Size by Market Sentiment")
plt.xlabel("Market Sentiment")
plt.ylabel("Average Trade Size (USD)")

plt.savefig("trade_size_by_sentiment.png", dpi=300, bbox_inches='tight')

plt.show()
```

**Insights:**

1. Traders used the largest average trade sizes during Fear conditions, while Extreme Greed showed the smallest average position sizes.
2. Trader risk-taking behavior changes significantly with market sentiment.

```python
# Analyze long vs short behavior by sentiment

long_short_sentiment = (
    merged_df.groupby(['classification', 'Side'])
    .size()
    .reset_index(name='count')
)

print(long_short_sentiment)
```

```python
# Visualize long vs short behavior by sentiment

plt.figure(figsize=(10,5))

sns.barplot(
    x='classification',
    y='count',
    hue='Side',
    data=long_short_sentiment
)

plt.title("Long vs Short Activity by Market Sentiment")
plt.xlabel("Market Sentiment")
plt.ylabel("Number of Trades")

plt.savefig("long_vs_short_sentiment.png", dpi=300, bbox_inches='tight')

plt.show()
```

**Insights:**

1. SELL activity slightly exceeded BUY activity across most sentiment conditions, especially during Extreme Greed and Greed markets.
2. Trader directional behavior changes with market psychology and perceived market opportunity.

```python
# Correlation analysis between numeric features

numeric_cols = merged_df.select_dtypes(include=np.number)

corr_matrix = numeric_cols.corr()

# Visualize correlation matrix

plt.figure(figsize=(12,8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")

plt.savefig("correlation_heatmap.png", dpi=300, bbox_inches='tight')

plt.show()
```

**Insights:**

1. Most trading variables showed weak correlation with Closed PnL, indicating that trader profitability is likely influenced by multiple complex factors rather than a single metric.
2. Trader performance is behaviorally complex and cannot be explained using simple linear relationships alone.

```python
# Compare PnL distribution across sentiment categories

plt.figure(figsize=(10,5))

sns.boxplot(
    x='classification',
    y='Closed PnL',
    data=merged_df
)

plt.title("PnL Distribution by Market Sentiment")
plt.xlabel("Market Sentiment")
plt.ylabel("Closed PnL")

plt.savefig("pnl_distribution_boxplot.png", dpi=300, bbox_inches='tight')

plt.show()
```

**Insights:**

1. PnL distributions across all sentiment categories show significant volatility and many extreme outliers, especially during Fear and Greed conditions.
2. Trader outcomes vary significantly under emotional market conditions.
3. Market sentiment strongly affects the volatility and dispersion of trading outcomes, not just average profitability.

```python
# Create trader activity segments

trade_counts = (
    merged_df.groupby('Account')
    .size()
    .reset_index(name='trade_count')
)

# Median trade count
threshold = trade_counts['trade_count'].median()

# Create activity segment
trade_counts['trader_type'] = np.where(
    trade_counts['trade_count'] >= threshold,
    'Frequent Trader',
    'Infrequent Trader'
)

# Preview
trade_counts.head()
```

```python
# Merge trader segments back into main dataset

merged_df = merged_df.merge(
    trade_counts[['Account', 'trader_type']],
    on='Account',
    how='left'
)

# Compare average PnL by trader type

segment_pnl = (
    merged_df.groupby('trader_type')['Closed PnL']
    .mean()
    .reset_index()
)

print(segment_pnl)
```

```python
# Visualize profitability by trader segment

plt.figure(figsize=(8,5))

sns.barplot(
    x='trader_type',
    y='Closed PnL',
    data=segment_pnl
)

plt.title("Average PnL by Trader Type")
plt.xlabel("Trader Type")
plt.ylabel("Average Closed PnL")

plt.savefig("profitability_by_trader_type.png", dpi=300, bbox_inches='tight')

plt.show()
```

**Insights:**

1. Infrequent traders achieved higher average profitability per trade compared to frequent traders.
2. Higher trading frequency does not automatically lead to better trading efficiency.
3. The average profit per trade of Infrequent Traders is higher but this doesn't mean that Infrequent traders make more total profit overall.

```python
# Compare average trade size by trader segment

segment_trade_size = (
    merged_df.groupby('trader_type')['Size USD']
    .mean()
    .reset_index()
)

print(segment_trade_size)
```

```python
# Visualize Avg. Trade size by trader segment

plt.figure(figsize=(8,5))

sns.barplot(
    x='trader_type',
    y='Size USD',
    data=segment_trade_size
)

plt.title("Average Trade Size by Trader Type")
plt.xlabel("Trader Type")
plt.ylabel("Average Trade Size(USD)")

plt.savefig("trade_size_by_trader_type.png", dpi=300, bbox_inches='tight')

plt.show()
```

**Insights:**

1. Frequent traders used larger average trade sizes compared to infrequent traders.
2. Frequent traders take larger trades but achieve lower average profitability per trade.
3. Higher risk-taking and larger trade sizes do not automatically translate into better trading performance.

```python
# Compare win rate by trader segment

segment_win_rate = (
    merged_df.groupby('trader_type')['is_profit']
    .mean()
    * 100
).reset_index()

# Rename column
segment_win_rate.columns = ['Trader Type', 'Win Rate %']

print(segment_win_rate)
```

```python
# Visualize win rate by trader segment

plt.figure(figsize=(8,5))

sns.barplot(
    x='Trader Type',
    y='Win Rate %',
    data=segment_win_rate
)

plt.title("Win Rate by Trader Type")
plt.xlabel("Trader Type")
plt.ylabel("Win Rate (%)")

plt.savefig("winrate_by_trader_type.png", dpi=300, bbox_inches='tight')

plt.show()
```

**Insights:**

1. Frequent traders achieved slightly higher win rates compared to infrequent traders, but the difference isn't much.
2. Trading frequency influences both consistency and profit efficiency in different ways.
3. The findings imply:
  *   frequent trading may improve win rate
  *   selective trading may improve profit quality

# Strategy Recommendations

**STRATEGY 1: Selective Trading During Greed Markets**

**Recommendation**

During Greed and Extreme Greed conditions, traders should prioritize selective high-conviction trades instead of excessive frequent trading.

**Analysis showed:**

* Infrequent traders achieved significantly higher average profitability during Greed conditions.
* Frequent traders had higher activity but lower profit efficiency.
* Extreme Greed produced the highest average profitability and win rates.

**This suggests:**

* Bullish markets reward patience and trade quality
* Overtrading during euphoric conditions may reduce efficiency
* Selective momentum participation performs better.

**STRATEGY 2 — Reduce Exposure During Fear Markets**

**Recommendation**

During Fear and Extreme Fear conditions, traders should reduce position sizes and apply stricter risk management.

**Analysis showed:**

* Fear markets had the largest average trade sizes.
* PnL distributions during Fear showed extreme volatility and large outliers.
* Win rates dropped during Extreme Fear conditions.

**This suggests:**

* Fear-driven markets create unstable trading environments
* Aggressive exposure increases downside risk
* Smaller position sizing may improve long-term consistency.

**STRATEGY 3 — Avoid Overtrading**

**Recommendation**

Higher trading frequency should not automatically be treated as a profitable strategy.

**Analysis showed:**

Frequent traders had:
* Larger trade sizes
* Higher win rates
* Lower average profitability per trade

**This suggests:**

* more trades ≠ better profitability
* excessive participation may reduce efficiency
* disciplined trade selection matters more than trade quantity

This project analyzed the relationship between Market sentiment and trader behavior on Hyperliquid using Fear & Greed sentiment data and historical trading activity.
